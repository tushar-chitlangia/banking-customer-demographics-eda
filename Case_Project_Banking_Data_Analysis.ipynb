```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style for clean, professional-looking charts
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
# Load the dataset
df = pd.read_csv('banking_data.csv')

print(f"Dataset Loaded! Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")
df.head()
```

```python
# Age Distribution of Clients
print(df['age'].describe())

plt.figure(figsize=(10, 5))
sns.histplot(df['age'], bins=30, kde=True, color='skyblue')
plt.title('Distribution of Client Age', fontsize=14)
plt.xlabel('Age (Years)', fontsize=12)
plt.ylabel('Number of Clients', fontsize=12)
plt.show()
```

```python
# Job Type Variation
job_counts = df['job'].value_counts()
job_percentages = df['job'].value_counts(normalize=True) * 100

# table summary
job_summary = pd.DataFrame({'Count': job_counts, 'Percentage (%)': job_percentages.round(2)})
print(job_summary)

plt.figure(figsize=(12, 6))
sns.countplot(data=df, y='job', order=job_counts.index, palette='viridis')
plt.title('Job Type Variation Among Clients', fontsize=14)
plt.xlabel('Number of Clients', fontsize=12)
plt.ylabel('Job Category', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# Marital Status & Education Levels

print("Marital Status Distribution")
marital_counts = df['marital'].value_counts()
print(df['marital'].value_counts(normalize=True).round(4) * 100)

plt.figure(figsize=(6, 6))
plt.pie(marital_counts, labels=marital_counts.index, autopct='%1.1f%%', colors=['#4f81bd', '#c0504d', '#9bbb59'])
plt.title('Marital Status Distribution', fontsize=14)
plt.show()

print("\nLevel of Education Among Clients")
print(df['education'].value_counts())

plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='education', order=df['education'].value_counts().index, palette='magma')
plt.title('Education Levels of Clients', fontsize=14)
plt.xlabel('Education Level', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.show()
```

```python
# Financial Profile Loans & Defaults
print("Proportion of Clients with Credit in Default")
print(df['default'].value_counts(normalize=True).round(4) * 100)

print("\nHow many clients have housing loans?")
print(df['housing'].value_counts())

print("\nHow many clients have personal loans?")
print(df['loan'].value_counts())

# Plotting comparisons side by side
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.countplot(data=df, x='housing', ax=axes[0], palette='Set2')
axes[0].set_title('Has Housing Loan?', fontsize=13)

sns.countplot(data=df, x='loan', ax=axes[1], palette='Set1')
axes[1].set_title('Has Personal Loan?', fontsize=13)
plt.show()
```

```python
# Average Yearly Balance
print("Average Yearly Balance Metrics (Euros)")
print(df['balance'].describe())

plt.figure(figsize=(10, 4))
sns.boxplot(x=df['balance'], color='lightgreen')
plt.title('Box Plot of Average Yearly Balance', fontsize=14)
plt.xlabel('Balance (Euros)', fontsize=12)
plt.show()
```

```python
print("Communication Types Used")
print(df['contact'].value_counts(normalize=True).round(4) * 100)

print("\nLast Contact Day of the Month Summary")
print(df['day'].describe())

print("\nLast Contact Month Distribution")
month_order = ['jan', 'feb', 'mar', 'apr', 'may', 'jun', 'jul', 'aug', 'sep', 'oct', 'nov', 'dec']
print(df['month'].value_counts().reindex(month_order).fillna(0))

# Months Timeline
plt.figure(figsize=(12, 5))
sns.countplot(data=df, x='month', order=month_order, palette='coolwarm')
plt.title('Variation of Outreach Across Months', fontsize=14)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Number of Contacts', fontsize=12)
plt.show()
```

```python
print("Last Contact Duration Metrics (Seconds)")
print(df['duration'].describe())

print("\nContacts Performed During Current Campaign")
print(df['campaign'].describe())

# Plot distribution of call lengths (cropped to 2000s for visibility)
plt.figure(figsize=(10, 5))
sns.histplot(df['duration'], bins=100, color='purple', kde=True)
plt.xlim(0, 2000)
plt.title('Distribution of Last Contact Duration (Up to 2000 Seconds)', fontsize=14)
plt.xlabel('Duration (Seconds)', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.show()
```

```python
print("Days Passed Since Previous Campaign (pdays)")
not_contacted = (df['pdays'] == -1).sum()
print(f"Clients never contacted before: {not_contacted} ({not_contacted/len(df)*100:.2f}%)")

print("\nContacts Performed Before Current Campaign")
print(df['previous'].describe())

print("\nOutcomes of Previous Marketing Campaigns")
print(df['poutcome'].value_counts(normalize=True).round(4) * 100)
```

```python
print("Term Deposit Subscription Distribution (y)")
target_counts = df['y'].value_counts()
target_pct = df['y'].value_counts(normalize=True) * 100

summary_y = pd.DataFrame({'Count': target_counts, 'Percentage (%)': target_pct.round(2)})
print(summary_y)

plt.figure(figsize=(6, 5))
sns.countplot(data=df, x='y', palette='Set2')
plt.title('Subscription to Term Deposit Results (Target Variable y)', fontsize=14)
plt.xlabel('Subscribed to Term Deposit?', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.show()
```
